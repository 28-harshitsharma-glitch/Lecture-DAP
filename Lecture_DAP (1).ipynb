```python
#Data type
print(df.dtypes)
```

```python
#Top 10 profit/loss product
df["Total_Cost_Lakh"] = (
    df["Marketing_Spend_Lakh"]
    +df["Content_Cost_Lakh"]
    +df["Salary_Cost_Lakh"]
    +df["Other_Cost_Lakh"]
)

df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total cost lakh"]

print(df[["Product", "Revenue_Lakh", "Total_Cost_Lakh", "Profit_Lakh"]].head(10))
```

```python
#Missing Value
print(df.isna().sum())
```
